# 1.DISTRIBUTION PLOTS

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

##### We use distribution plots to visualize the distribution of quantative data.
##### Let's discuss some plots that allow us to visulaize the distribution of a data set. These plots are :
##### -> distplot/histplot
##### -> jointplot
##### -> pairplot
##### -> rugplot

```python
df = sns.load_dataset('tips')
df
```

```python
df['size'].unique()
```

### Distplot/histplot

```python
plt.subplot(1,2,1)
sns.histplot(df['total_bill'], bins = 20, kde = True)
plt.subplot(1,2,2)
sns.histplot(df['tip'], bins = 20, kde = True)

```

### Jointplot

```python
sns.jointplot(x= 'total_bill', y = 'tip' , data = df, kind = 'scatter') 
```

```python
sns.jointplot(x= 'total_bill', y = 'tip' , data = df, kind = 'reg') 
```

```python
sns.jointplot(x= 'total_bill', y = 'tip' , data = df, kind = 'hex') 
```

### Pairplot

```python
sns.pairplot(df,hue = 'size' , palette = 'rainbow')
```

### Rugplot

```python
sns.rugplot(df)
```

# 2. CATEGORICAL DATA PLOTS

##### Let's discuss using seaborn to plot categorical data!. These are a few main plot types are for this:
##### -> boxplot
##### -> violinplot
##### -> stripplot
##### -> barplot
##### -> swarmplot
##### -> countplot
##### Lets go through examples of each!

```python
df = sns.load_dataset('tips')
df
```

```python
#### countplot
```

```python
sns.countplot(x = df['sex'] , hue = df['smoker'])
```

```python
#### Barplot
```

```python
sns.barplot( x = df['sex'], y = df['total_bill'])
```

```python
import numpy as np
sns.barplot( x = df['sex'], y = df['total_bill'] , estimator = np.sum )
```

```python
sns.barplot( x = df['sex'], y = df['tip'] , estimator = np.sum )
```

#### Boxplot

```python
sns.boxplot(x = 'tip' , y = 'day' , data = df , palette = 'rainbow')
```

#### Violinplot

```python
sns.violinplot(x = 'day' , y = 'total_bill' , data = df)
```

#### Stripplot

```python
sns.stripplot(x = 'day' , y = 'total_bill' , data = df)
```

#### Swarmplot

```python
sns.swarmplot( x = 'day' , y = 'total_bill' , data = df)
```

```python
sns.violinplot(x = 'tip' , y = 'day' , data = df)
sns.swarmplot( x = 'tip' , y = 'day' , data = df)
```
