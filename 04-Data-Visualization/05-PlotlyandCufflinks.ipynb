```python
import plotly.express as px

fig = px.histogram(
    tips,
    x='total_bill',
    title='Total Bill Distribution'
)

fig.show()
```

```python
import plotly.express as px

avg_tip = tips.groupby('day')['tip'].mean().reset_index()

fig = px.bar(
    avg_tip,
    x='day',
    y='tip'
)

fig.show()
```

```python
import plotly.express as px

fig = px.line(
    tips,
    y='total_bill',
    title='Total Bill Distribution'
)

fig.show()
```
