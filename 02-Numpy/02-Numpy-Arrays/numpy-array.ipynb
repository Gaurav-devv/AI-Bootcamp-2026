RANDOM GENERATION FUNCTION

```python
np.random.rand(10) #The value will come between 0 and 1  this is called Normalization.
```

```python
np.random.randn(10) # The value will be generated betwwen -3 to +3, called as Standrization
```

```python
np.random.randint(10,20,10)  #(start, end, numbers b/w start and end)
```
