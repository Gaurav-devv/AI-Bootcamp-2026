```python
import numpy as np
```

NUMPY INDEXING AND SLICING OF VECTORS

```python
arr=np.arange(11,21)
arr
```

```python
arr[6]   #using index value to print the required number
```

```python
arr_slice = arr[3:9:2]
arr_slice
```

NUMPY INDEXING AND SLICING OF MATRIX

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
arr[5] #Finding the array of the row by its index
```

```python
arr[0,4]
```

```python
slice = arr[0:2,1:3] #slicing a portion from matrix
slice
```

```python
slice = arr[3:6,3:5]
slice
```

BOOLEAN INDEXING

```python
arr = np.arange(11,21)
arr
```

```python
bool_index = arr%2 == 0
bool_index
```

```python
arr = arr[bool_index]
arr
```
