```python
import numpy as np

#Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],  #Student 1
    [19, 92, 88],  #Student 2
    [17, 76, 95],  #Student 3
    [18, 65, 70],  #Student 4
    [20, 90, 85]  ##Student 5  
])
```

```python
#Get the shape of the matrix.
data.shape
```

```python
#Find the average  age of students
np.mean(data[:,0])
```

```python
#Extract Math marks of all students.
data[:,1]
```

```python
#Find the highest Science mark.
np.max(data[:,2])
```

```python
#Get details of the student who scored more than 90 in Maath.

data[data[:,1] > 90]
```

```python
#Increase Math marks of all students by 5.

data[:,1] += 5
data
```

```python
#Find how many students are younger than 19.

len(data[data[:,0] < 19])
```

```python
#Calculate the average marks in each subject(Column-wise mean)

np.mean(data[:,1:], axis = 0)
```

```python
#Get data of students who scored at least 80 in both sbjects.

data[(data[:,1] >= 80) & (data[:,2] >= 80)]
```

```python
#Replace all science marks < 75 with 0.

data[:,2][data[:,2] < 75] = 0
data
```
